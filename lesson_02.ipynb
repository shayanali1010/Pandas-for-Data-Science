## Data Cleaning and Preprocessing:

```python
import pandas as pd
```

### Handling Missing Data:

```python
data = pd.Series([11, None, 45, 78, None])

data.dropna()  # remove all entries (rows) where value is NaN
data.fillna(5)  # replace NaN with fixed value
data.fillna(data.mean())  # replace NaN with mean
data.fillna(data.min())  # replace NaN with minimum value
data.fillna(data.max())  # replace NaN with maximum value
```

### Remove Duplicates:

```python
s1 = pd.Series([1,2,1,4])
s2 = pd.Series([1,23,1,45])

data_1 = pd.DataFrame({
    'C1': s1,
    'C2': s2
})

data_1.drop_duplicates() # removes rows where ALL column values are exactly the same
```

### Replace Values:

```python
data_1.replace(1, 233) # replace all occurrences of 1 with 233 (entire DataFrame, not row/column specific)
```

```python
data_1['C1'] = data_1['C1'].astype(float)  # convert C1 column data type to float
data_1
```

### String Operations:

```python
data_1['C3'] = ['yes', 'no', 'or', 'and']  # create new column C3 with string values
# data_1['C3'].str.lower()  # convert all strings in C3 to lowercase
data_1['C3'].str.upper()  # convert all strings in C3 to uppercase (returns new Series)
```

## Data Aggregation and Grouping:

### GroupBy Example:

```python
s1 = pd.Series(['A', 'B', 'A', 'B', 'C'])
s2 = pd.Series([11, 23, 53, 55, 34])

# Create DataFrame
data_2 = pd.DataFrame({
    'Name': s1,
    'lemonades': s2
})

# groupby('Name') groups rows having the same Name together
data_2.groupby('Name').sum()    # Sum of values
data_2.groupby('Name').mean()   # Average of values
data_2.groupby('Name').max()    # Highest value
data_2.groupby('Name').min()    # Lowest value

```

### Pivot Tables:

```python
# pivot_table() groups rows by 'Name' and summarizes the 'lemonades' numbers
# If a Name appears multiple times, its lemonade values are combined
# By default, pivot_table() calculates the average (mean) of those numbers

# By default mean
data_2.pivot_table(values='lemonades', index='Name')

# Sum of values
data_2.pivot_table(values='lemonades', index='Name', aggfunc='sum')

# Maximum value
data_2.pivot_table(values='lemonades', index='Name', aggfunc='max')

# Minimum value
data_2.pivot_table(values='lemonades', index='Name', aggfunc='min')

# Count how many values are in each group
data_2.pivot_table(values='lemonades', index='Name', aggfunc='count')
```

### Crosstab:

```python
# crosstab() counts how many times each combination of 'Name' and 'lemonades' appears
# Rows represent unique Names and columns represent lemonade values
# The numbers in the table show the frequency/count of occurrences

pd.crosstab(data_2['Name'], data_2['lemonades'])
```

## Merging, Joining and Concatinating:

### Concatinating:

```python
s1 = pd.Series(['A', 'B', 'A', 'B', 'D'])
s2 = pd.Series([11, 22, 55, 66, 33])

data_3 = pd.DataFrame({
    'Name': s1,
    'lemon': s2
})

# concat() stacks DataFrames row-wise; matching columns are combined, non-matching columns are filled with NaN only where data is missing
pd.concat([data_2, data_3])
```

### Merge:

```python
# Merge data_2 and data_3 using "Name" as the key column
# Only rows with EXACT matching values in "Name" in BOTH DataFrames are kept
# If names are different or missing in either DataFrame, those rows are excluded (inner join behavior)
pd.merge(data_2, data_3, on='Name')
```

### Join on Index:

```python
# Join data_2 with data_3 by matching data_2["Name"] with data_3 index
# Default is LEFT join → keeps all rows from data_2, missing matches become NaN
# Unlike merge(), join uses index-based matching (after set_index) instead of column-to-column matching

data_2.join(data_3.set_index('Name'), on='Name')
```
